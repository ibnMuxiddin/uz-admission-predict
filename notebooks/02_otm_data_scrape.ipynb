
# Mandat saytidan ma'lumotlarni yig'ish (Scraping)
Bu yerda `https://mandat.uzbmb.uz/` saytidan:
- Hudud: Toshkent shahri (14)
- OTM: Toshkent davlat iqtisodiyot universiteti (324)
- Yo'nalish: Davlat auditi (60411500)
- Ta'lim tili: O'zbekcha (1)
- Ta'lim shakli: Kunduzgi (1)

ushbu filtrlarga mos keluvchi abituriyentlar ro'yxatidan statusi "To'lov shartnoma" bo'lgan abituriyentlarni sanash kodi yozilgan.


```python

import requests
from bs4 import BeautifulSoup

def count_contract():
    base_url = "https://mandat.uzbmb.uz/Mandat2025/Paginate"
    page = 1
    total_contract = 0
    print("Ma'lumotlar yuklanmoqda...")
    while True:
        params = {
            'pageNumber': page,
            'pageSize': 100, 
            'region': 14,
            'univer': 324,
            'faculty': 60411500,
            'edlang': 1,
            'edtype': 1
        }
        r = requests.get(base_url, params=params)
        soup = BeautifulSoup(r.text, 'html.parser')
        items = soup.find_all('div', class_='m3-rank-item')
        
        if not items:
            break
            
        for item in items:
            classes = item.get('class', [])
            if 'm3-rank--contract' in classes:
                total_contract += 1
            
        # Check if next page exists
        pag = soup.find('ul', class_='pagination')
        has_next = False
        if pag:
            for btn in pag.find_all('button'):
                if 'Keyingi' in btn.text:
                    has_next = True
                    break
        
        if not has_next:
            break
            
        page += 1
        
    print(f"To'lov shartnoma asosida qabul qilinganlar soni: {total_contract}")
    return total_contract

total_count = count_contract()
```
